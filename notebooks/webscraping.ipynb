b. Now, use what you did for the first title, but extract the job title for all jobs on this page. Store the results in a list.

```python
job_titles = []
elements = fakepython.find_all('h2', class_="title is-5")
for element in elements:
    job_titles.append(element.text)

print(job_titles)
```

c. Finally, extract the companies, locations, and posting dates for each job. For example, the first job has a company of "Payne, Roberts and Davis", a location of "Stewartbury, AA", and a posting date of "2021-04-08". Ensure that the text that you extract is clean, meaning no extra spaces or other characters at the beginning or end.

```python
job_companies = []
elements = fakepython.find_all('h3', class_="subtitle is-6 company")
for element in elements:
    job_companies.append(element.text)

print(job_companies)
```

```python
job_locations = []
elements = fakepython.find_all('p', class_="location")
for element in elements:
    job_locations.append(element.text)
    job_locations = [x.replace(' ', '') for x in job_locations]
    job_locations = [x.replace('\n', '') for x in job_locations]
    job_locations = [x.replace(',', ', ') for x in job_locations]

print(job_locations)
```

```python
job_dates = []
elements = fakepython.find_all('p', class_="is-small has-text-grey")
for element in elements:
    job_dates.append(element.text)
    job_dates = [x.replace('\n', '') for x in job_dates]

print(job_dates)
```

d. Take the lists that you have created and combine them into a pandas DataFrame.

```python
jobs = pd.DataFrame({'title': job_titles, 'company': job_companies, 'location': job_locations, 'date': job_dates})
jobs
```

```python
soup_urls = []
links = fakepython.find_all("a")
for link in links:
    if link.text == 'Apply':
        soup_url = link.get("href")
        soup_urls.append(soup_url)

print(soup_urls)
```

Finally, we want to get the job description text for each job.

a. Start by looking at the page for the first job, https://realpython.github.io/fake-jobs/jobs/senior-python-developer-0.html. Using BeautifulSoup, extract the job description paragraph.

```python
# fakepython = BS(fp.text)
# print(fakepython.prettify())
```

```python
descriptions = []
for url in urls:
    response = requests.get(url)
    soup = BS(response.text, "html.parser")

    descs = soup.find_all('div', class_="content")
    for x in descs:
        desc = x.split('Location').str(0)
        descriptions.append(desc)

print(descriptions)
# print(soup.prettify())
```
